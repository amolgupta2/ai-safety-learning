### Initializing Different Types of Arrays

````python
import numpy as np
import sys
````

````python
# All 0s Matrix
sukablyat = np.zeros((2, 5))
print(sukablyat)
````

````python
# All 1s Matrix with Data Type int 8

np.ones((2, 2, 4), dtype="int8")
````

````python
# Initializing Matrices with other numbers
np.full((2, 2), 45, dtype="float64")
````

````python
# Initializing matrices with other numbers (full_like)

np.full_like(sukablyat, 67)
````

````python
# Initializing a matrix of random decimal numbers
np.random.rand(2, 2)
np.random.random_sample(sukablyat.shape)
````

````python
# Initializing a matrix of random integer values
np.random.randint(90, 100, size=(2, 2))
````

````python
# Identity Matrix

np.identity(4)
````

````python
# Repeat an array

sukablyat = np.array([[1, 2, 3]])
print(np.repeat(sukablyat, 3, axis = 0))
````

````python
'''
My thinking process:
2-Dimensional array with 5 rows and 5 columns
First and last rows are just 1s repeated.
Second and fourth rows have both 1s and 0s.
Middle row is like the fourth and second with a 9 in the second index.
'''

sukablyat = np.ones((5, 5), dtype='int8')
sukablyat[1:4, 1:4] = np.zeros((3, 3))
sukablyat[2, 2] = 9
print(sukablyat)
````

### Be careful when copying arrays!

````python
a = np.array([1, 2, 3])
k = a
print(k)

k[0] = 99
print(a)
# So both k and a point to the same array in space, not an actual copy
````

````python
# Let's try this again with .copy()

a = np.array([1, 2, 3])
k = a.copy()
k[0] = 99

print(k)
print(a)
````
