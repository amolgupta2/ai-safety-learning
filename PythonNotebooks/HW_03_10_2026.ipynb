Problem 1 — Slicing and Indexing
Given this matrix:

```
pythonA = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])
```

Extract the following without creating a new array manually:
* The second row
* The third column
* The value 50
* The bottom-right 2x2 submatrix

````python
# Problem 2 - My Attempt

pythonA = np.array([[10, 20, 30],
                    [40, 50, 60],
                    [70, 80, 90]])

# The second row
print(pythonA[1, :])

# The third column
print(pythonA[:, 2])

# The value 50
print(pythonA[1, 1])

# The bottom-right 2x2 submatrix
print(pythonA[1:, 1:])
````
