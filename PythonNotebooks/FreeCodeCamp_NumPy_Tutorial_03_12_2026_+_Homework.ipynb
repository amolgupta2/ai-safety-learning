## Reorganizing Arrays

````python
before = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print(before)

after = before.reshape((8, 1))
print(after)

afterthat = after.reshape((2, 2, 2))
print(afterthat)

evenmore = before.reshape(4, 1, 1, 2)
print(evenmore)
````

````python
# Vertically stacking vectors/matrices

v1 = np.array([1, 2, 3, 4])
v2 = np.array([5, 6, 7, 8])

np.vstack([v1, v2, v2, v2])
````

````python
# Horizontal Stacks
h1 = np.array([[1], [2], [3], [4]])
h2 = np.array([[5], [6], [7], [8]])

np.hstack([v1, v2])

h1 = np.full((2, 4), 69)
h2 = np.full((2, 2), 67)

np.hstack([h1, h2])
````

###Problem 3 — Linear Algebra
Create these two matrices manually using np.array:
```
A = [[2, 1],
     [0, 3]]

b = [[1, 4],
     [2, 0]]
```
Then compute:
- A * B
- B * A
- Are they the same? Print both and write a one line comment on what this tells you about matrix multiplication
- What you discover in this problem connects directly back to what you were just doing by hand with 3B1B.

````python
A = np.array([[2, 1],
              [0, 3]])

B = np.array([[1, 4],
              [2, 0]])

print("Result of A X B")
print(np.matmul(A, B))

print("\nResult of B X A")
print(np.matmul(B, A))

'''
The result of (A X B) != (B X A) which means Matrix Multiplication is not
Commutative. However, this is expected as we are performing 2 distinct linear
transformations, and doing them in different orders is going to change the way
the basis vectors will transform
'''
````

### Problem 5 - Bringing it all Together
Create a (4, 4) matrix of random integers between 1 and 50. Then:
- Compute the sum of every row
- Compute the sum of every column
- Find the overall mean of the entire matrix
- Reshape it into a (2, 8) matrix and compute the mean of each row
Do all of this without any loops. NumPy only

````python
p5 = np.random.randint(1, 51, size=(4, 4))

# Sum of every column
print(np.sum(p5, axis = 0))

# Sum of every row
print(np.sum(p5, axis = 1)) # Sum of Rows

# Mean of overall mean of entire matrix
print(np.mean(p5))

# Well obviously! NumPy is Goated!
p5 = p5.reshape(2, 8)
print(np.mean(p5, axis = 1))
````
