# Lesson 4: Basic Operators (+, -, *, /, //, %)

Python follows the Order of Operations just like normal Mathematics:

1. Parentheses: ()
2. Exponents: **
3. Multiply/Divide/Modulo/Divide w/out remainder: *, /, %, //
4. Add/Subtract: +, -

Python Specific Operators: % and //

1. Modulo (%): When dividing 2 numbers, it finds the remainder
   - Example: 99 % 10 = 9

2. Floor Division (//): When dividing 2 numbers, it removes whatever remainder would be there
    - Example: 99 // 10 = 9

# Lesson 6: Loops

2 Types of Loops: For and While

For loops continue for a set range with a possible bottom, top, and interator:
- For Example:
```
for x in range(0, 10, 2):
  print(x) # Will print every even number in (0-8)
```

While loops continue as long as a Boolean condition is still true. Example:
```
while (x < 10):
  print("Wait")
  x += 1
else:
  print("You may now cross the crosswalk")
```

But also within loops are 2 statements that can change the behavior of loops:
1. break
2. continue

Break statements "break" the loop entirely and move on to the remeanining code after the loop

Continue statements "continue" skip the rest of the code in the loop and start from the beginning again

Finally, there are 'else' blocks within loops as well:
1. These else blocks are triggered only when the loop conditions fail (which will always be at the end)
2. If there is a break statement triggered before the else statement, the else statement code does NOT get executed
3. If there is a continue statement triggered before the else statement, the else statement code does get executed once the loop fails
